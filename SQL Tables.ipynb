## 1.	SQL Database – lewis_customers.db
- **Customers:** customer info (id, gender, age, region, loyalty, account type, join date)  
- **Transactions:** purchases (id, customer, store, date, category, amount)  
- **Stores:** store info (id, region, manager, target)


```python
import sqlite3
import os
import pandas as pd
import numpy as np
from faker import Faker
from datetime import datetime, timedelta
import random, os
```

```python
db_path = r"C:/Users/SD 1/Downloads/Data Set Creation (1)/src/db/lewis_customers.db"
```

```python
conn = sqlite3.connect(db_path)
```

```python
fake =Faker()
np.random.seed(42)
random.seed(42)
#creates a fake data generator ,and making sure that the randomness is repeatable 
```

### Customers Table 


```python
customers = []
regions = ['Western Cape','Eastern Cape','KwaZulu-Natal', 'Gauteng','Limpopo']
account_types =['Credit', 'Cash', 'Bank Accounts']

```

```python
for cid in range(1,201):
    join_date = fake.date_between(start_date ='-3y',end_date ='today')
    customers.append((
    cid, 
    fake.random_element(elements=('Male','Female')),
    random.randint(20,65),
    random.choice(regions),
    random.randint(20,100),
    random.choice(account_types),
    join_date 
    ))
```

```python
customer_df = pd.DataFrame(customers,
                           columns= [ "customer_id", "Gender", "Age", "Region", "Loyalty Score", "Account Type", "Join Date"])
```

### Stores Table


```python
stores = [] 
for sid in range(1,6):
    stores.append((
        sid,
        random.choice(regions),
        fake.name(),
        random.randint(200000,500000)
    ))
```

```python
store_df = pd.DataFrame(stores ,
                        columns = ['store_id','region','manager','target'])
```

### Transaction Table

```python
transactions=[]
categories = ['Lounge','Bedroom','Electronics','Appliances','Decor','Dining']
```

```python
for tid in range(1, 801):
    cust_id = random.randint(1, 200)
    store_id = random.randint(1, 5)
    category = random.choice(categories)
    date = fake.date_between(start_date="-6m", end_date="today")
    total_spent = random.randint(800, 20000)
    transactions.append((tid, cust_id, store_id, date, category, total_spent))

transactions_df = pd.DataFrame(transactions, columns=[
    "transaction_id", "customer_id", "store_id", "date", "category", "total_spent"
])
```

```python
con = sqlite3.connect(db_path)
customer_df.to_sql("Customers", con, if_exists="replace", index=False)
store_df.to_sql("Stores", con, if_exists="replace", index=False)
transactions_df.to_sql("Transactions", con, if_exists="replace", index=False)
con.close()

print("✅ Database created:", db_path)
```
